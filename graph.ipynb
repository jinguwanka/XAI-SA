```python
import pandas as pd
import nltk
from nltk.corpus import stopwords


# 加载葡萄牙语 stopwords
stop_words = set(stopwords.words('portuguese'))

# 读取 CSV 文件
df = pd.read_csv("token_attributions_full.csv")

# 预处理：去掉特殊 token，小写化
df = df[~df["token"].isin(["[PAD]", "[CLS]", "[SEP]"])]
df["token"] = df["token"].str.lower()

# 去除停用词
df = df[~df["token"].isin(stop_words)]

# 设置 attribution score 阈值
score_threshold = 0.15

# 存储结果
high_freq_tokens = {}

# 对正向 / 负向分别处理
for label in ["positive", "negative"]:
    sub_df = df[df["label"] == label]

    # 统计词频和平均 attribution score
    token_stats = sub_df.groupby("token").agg(
        freq=("token", "count"),
        avg_score=("attribution_score", "mean")
    ).reset_index()

    # 只保留 freq 前20% + avg_score >= 0.15 的词
    freq_threshold = token_stats["freq"].quantile(0.8)

    filtered = token_stats[
        (token_stats["freq"] >= freq_threshold) &
        (token_stats["avg_score"] >= score_threshold)
    ]

    # 排序：按 freq 从高到低
    filtered = filtered.sort_values(by="freq", ascending=False)

    high_freq_tokens[label] = filtered

# 打印结果
print("🔷 Positive 类别高频词（score >= 0.15, 按频率排序）：")
print(high_freq_tokens["positive"].head(20))

print("\n🔶 Negative 类别高频词（score >= 0.15, 按频率排序）：")
print(high_freq_tokens["negative"].head(20))
```

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

# 创建词云函数
def generate_wordcloud(data, title):
    # 将 token: freq 转换成 dict
    word_freq = dict(zip(data["token"], data["freq"]))

    # 创建词云对象
    wc = WordCloud(
        width=800,
        height=400,
        background_color="white",
        max_words=100,
        font_path=None  # 可指定 ttf 字体路径，如果支持葡语带重音词的话
    ).generate_from_frequencies(word_freq)

    # 绘图
    plt.figure(figsize=(10, 5))
    plt.imshow(wc, interpolation="bilinear")
    plt.axis("off")
    plt.title(title, fontsize=20)
    plt.tight_layout()
    plt.show()

# 生成 Positive 类别词云
generate_wordcloud(high_freq_tokens["positive"], "")

# 生成 Negative 类别词云
generate_wordcloud(high_freq_tokens["negative"], "")
```

```python
import matplotlib.pyplot as plt
pt_to_en = {
    "produto": "product",
    "prazo": "deadline",
    "qualidade": "quality",
    "bom": "good",
    "antes": "before",
    "chegou": "arrived",
    "entrega": "delivery",
    "recebi": "received",
    "bem": "good",
    "tudo": "all",
    "entregue": "delivered",
    "negação": "negation",
    "veio": "came",
    "porém": "but",
    "aguardo": "await",
}
# 横向条形图函数
def plot_horizontal_top_tokens(pt_to_en, data, label_name, top_n=10):
    top_data = data.head(top_n).sort_values(by="freq")  # 为了从下往上显示，先排序
    def translate(token):
        en = pt_to_en.get(token, "")
        return f"{token} \n({en})" if en else token
    
    top_data["token"] = top_data["token"].apply(translate)
    plt.figure(figsize=(10, 8))
    bars = plt.barh(top_data["token"], top_data["freq"], color="#4a90e2" if label_name == "Positive" else "#e24a4a")
    plt.title(f"Top {top_n} Frequent Words - {label_name} Sentiment", fontsize=16)
    plt.xlabel("Frequency", fontsize=14)
    plt.ylabel("Word", fontsize=14)
    plt.yticks(fontsize=14)
    plt.xticks(fontsize=14)
    plt.tight_layout()

    # 显示数值标签
    for bar in bars:
        width = bar.get_width()
        plt.text(width + 0.5, bar.get_y() + bar.get_height()/2,
                 f"{int(width)}", va='center', fontsize=12)

    plt.show()



# 正负类分别绘图
plot_horizontal_top_tokens(pt_to_en, high_freq_tokens["positive"], label_name="Positive", top_n=10)
plot_horizontal_top_tokens(pt_to_en, high_freq_tokens["negative"], label_name="Negative", top_n=10)

```
